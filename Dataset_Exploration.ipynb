```python
import pandas as pd

#load the dataset
df=pd.read_csv("/content/sample_data/Dataset.csv")
df.head()
```

```python
#Rows and Columns
rows,columns=df.shape

print( "The number of rows is:",rows)
print("The number of columns is: ",columns)
```

```python
#Datatypes found in the dataset
print("\nDatatypes:")
print(df.dtypes)
```

```python
#Check for missing values
print("\n Missing value per column:")
print(df.isnull().sum())

total_missing=df.isnull().sum().sum()

if total_missing==0:
  print("\nThe dataset is complete. (No missing values)")
else:
  print("\nThe dataset contains ",total_missing," missing values")
```

```python
#creating dataset sample

first_15=df.head(15) #stores first the first 15 rows
last_20=df.tail(20)  #stores the last 20 rows

#merging the first 15 and last 20 rows
df.sample= pd.concat([first_15,last_20])

#display the sample dataset
print("/Sample Dataframe:")
display(df.sample)
```
